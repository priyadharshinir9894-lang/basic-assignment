```python
from pyspark.sql import SparkSession

# Creates a session or reuses an existing one smoothly
spark = SparkSession.builder \
    .appName("RDDExample") \
    .master("local") \
    .getOrCreate()

# Get the underlying SparkContext to work with RDDs
sc = spark.sparkContext
```

```python
from pyspark import SparkContext


numbers = [5, 12, 7, 20, 3, 16, 9, 24, 10, 1]
rdd = sc.parallelize(numbers)

filtered_rdd = rdd.filter(lambda x: x > 10)
result_rdd = filtered_rdd.map(lambda x: x * 2)

print(result_rdd.collect())
```

```python
from pyspark import SparkContext


student_marks = [
    ("Arun", 78),
    ("Divya", 92),
    ("Kavin", 45),
    ("Meena", 66),
    ("Ravi", 35),
    ("Priya", 88)
]

rdd = sc.parallelize(student_marks)

filtered_rdd = rdd.filter(lambda x: x[1] >= 50)

result_rdd = filtered_rdd.map(lambda x: f"{x[0]} scored {x[1]}")

print(result_rdd.collect())

sc.stop()
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("EmployeeDetails").getOrCreate()

employee_data = [
    (1, "Arun", "Male", 24, "Chennai", 28000),
    (2, "Divya", "Female", 27, "Bengaluru", 42000),
    (3, "Kavin", "Male", 22, "Chennai", 25000),
    (4, "Meena", "Female", 30, "Hyderabad", 52000),
    (5, "Ravi", "Male", 28, "Bengaluru", 38000),
    (6, "Priya", "Female", 25, "Chennai", 46000)
]

columns = ["id", "name", "gender", "age", "city", "salary"]

df = spark.createDataFrame(data=employee_data, schema=columns)

df.show()

df.select("name", "city", "salary").show()

df.filter(df.salary > 40000).show()

spark.stop()
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = SparkSession.builder \
    .appName("ProductData") \
    .getOrCreate()

product_data = [
    (101, "Notebook", 40, 5),
    (102, "Pen", 10, 20),
    (103, "Pencil", 5, 30),
    (104, "School Bag", 650, 3),
    (105, "Eraser", 8, 15)
]

columns = ["product_id", "product_name", "price", "quantity"]

df = spark.createDataFrame(product_data, columns)

df.show()

df = df.withColumn("total_value", col("price") * col("quantity"))

df.show()

spark.stop()
```

```python

from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = SparkSession.builder \
    .appName("MovieData") \
    .getOrCreate()

movie_data = [
    ("Leo", "Action", 250, 7.8),
    ("Hi Nanna", "Drama", 75, 8.2),
    ("Jailer", "Action", 200, 7.1),
    ("Good Night", "Comedy", 30, 8.0),
    ("Dragon", "Comedy", 100, 7.5),
    ("Amaran", "Drama", 150, 8.3)
]

columns = [
    "movie_name", "category",
    "collection_crore", "rating"
]

df = spark.createDataFrame(movie_data, columns)

df.select("movie_name", "category", "rating").show()

filtered_df = df.filter(col("rating") >= 8.0)

final_df = filtered_df.withColumn(
    "collection_with_10_percent_bonus",
    col("collection_crore") * 1.10
)

final_df.show()

spark.stop()
```
